```python
# Scraping data dari Google Play Store menggunakan library google-play-scraper
from google_play_scraper import app, Sort, reviews

result, continuation_token = reviews(
    'noice.app',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=15000, 
    filter_score_with=None
)

print(f"Total data: {len(result)}")
```

```python
# Konversi hasil scraping menjadi DataFrame dan menyimpannya ke file CSV
import pandas as pd

df = pd.DataFrame(result)
df.to_csv('noice_app_reviews.csv', index=False, encoding='utf-8')

print("Data berhasil disimpan")
```
